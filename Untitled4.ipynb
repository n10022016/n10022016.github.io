```python
# CPAN 112 - Income Tax Program
# This program asks the user for their name and annual gross pay,
# calculates the federal income tax using 2018 tax brackets,
# identifies the tax bracket, and prints a clean formatted receipt.

# Ask for user input
name = input("Enter your name: ")
salary = float(input("Enter your annual gross pay: "))

# Determine tax and bracket
if salary <= 46605:
    tax = salary * 0.15
    bracket = "15% (Income $46,605 or less)"

elif salary <= 93208:
    tax = (46605 * 0.15) + ((salary - 46605) * 0.205)
    bracket = "20.5% (Income over $46,605 up to $93,208)"

elif salary <= 144489:
    tax = (46605 * 0.15) + ((93208 - 46605) * 0.205) + ((salary - 93208) * 0.26)
    bracket = "26% (Income over $93,208 up to $144,489)"

elif salary <= 205842:
    tax = (46605 * 0.15) + ((93208 - 46605) * 0.205) + ((144489 - 93208) * 0.26) + ((salary - 144489) * 0.29)
    bracket = "29% (Income over $144,489 up to $205,842)"

else:
    tax = (46605 * 0.15) + ((93208 - 46605) * 0.205) + ((144489 - 93208) * 0.26) + ((205842 - 144489) * 0.29) + ((salary - 205842) * 0.33)
    bracket = "33% (Income over $205,842)"

# Round tax to 2 decimals
tax = round(tax, 2)

# Print formatted receipt
print("\n------------------------------")
print("          TAX SUMMARY")
print("------------------------------")
print("Name: ", name)
print("Annual Salary: $", format(salary, ".2f"))
print("Tax Bracket: ", bracket)
print("Federal Income Tax Owed: $", format(tax, ".2f"))
print("------------------------------")
```
